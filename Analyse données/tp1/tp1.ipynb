Q.3 Afficher graphiquement le résultat du clustering avec deux couleurs/symboles à l’aide de
kmeans.labels_ . Etudier les différences avec la “vraie” classification. adjusted_rand_score
du module metrics de la bibliothéque sklearn permet de calculer l’erreur entre la classification
obtenue et la classification “vraie”. Etudier l’aide pour comprendre la mesure.


```python
mean = np.array ([2, 2])
cov = np.array([[2, 0], [0, 2]])
A = (np.random.multivariate_normal(mean, cov, 128))

mean = np.array ([-4, -4])
cov = np.array([[6, 0], [0, 6]])
B = (np.random.multivariate_normal(mean, cov, 128))

data = np.concatenate((A,B), axis=0)
kmeans = KMeans (n_clusters = 2, n_init = 1, init = "k-means++")
kmeans.fit(data)

plt.scatter(data[:,0],data[:,1], c=kmeans.labels_, label="Cluster",)
plt.legend()
plt.show()
```

Les différence avec la vrai classification sont que les points présent sur le graphique sont répartit de deux coté ont des couleur qui peuvent différer car l'utilisatoin de kmeans a calculer 2 centre et les points sont de la couleur du centre le plus proche d'eux.

Q. 4 Utiliser la fonction adjusted_rand_score pour calculer le taux d’erreur.


```python
from sklearn.metrics import adjusted_rand_score
res = np.concatenate((np.zeros(128), np.ones(128)))
tauxErreur = (1-adjusted_rand_score(res,kmeans.labels_))*100
print("le taux d'érreur est de : ",tauxErreur,"%")
```

Le taux d'érreur est de 15% car des points sur le graphique ont changés de couleur car ils étaient plus proche d'un centre qui a était calculé.

### 1.2.2 Choix du "bon" nombre de clusters
Q.1 Séparer les données data en K = 2, 3, 4, 5, 6 clusters à l’aide de la fonction KMeans

```python
mean = np.array ([2, 2])
cov = np.array([[2, 0], [0, 2]])
A = (np.random.multivariate_normal(mean, cov, 128))

mean = np.array ([-4, -4])
cov = np.array([[6, 0], [0, 6]])
B = (np.random.multivariate_normal(mean, cov, 128))

data = np.concatenate((A,B), axis=0)
kmeans2 = KMeans (n_clusters = 2, n_init = 1, init = "k-means++")
kmeans2.fit(data)

kmeans3 = KMeans (n_clusters = 3, n_init = 1, init = "k-means++")
kmeans3.fit(data)

kmeans4 = KMeans (n_clusters = 4, n_init = 1, init = "k-means++")
kmeans4.fit(data)

kmeans5 = KMeans (n_clusters = 5, n_init = 1, init = "k-means++")
kmeans5.fit(data)

kmeans6 = KMeans (n_clusters = 6, n_init = 1, init = "k-means++")
kmeans6.fit(data)

plt.scatter(data[:,0],data[:,1], c=kmeans6.labels_, label="Cluster",)
plt.legend()
plt.show()
```

Q.2 Calculer les largeurs de silhouettes des observations et le coefficient de silhouette :

```python
from sklearn.metrics import silhouette_samples, silhouette_score

for K in range(2,7):
    kmeans = KMeans(n_clusters=K, n_init=1, init="k-means++").fit(data)
    cluster_labels = kmeans.labels_
    silhouette_avg = silhouette_score(data, cluster_labels)
    
    print("Pour le nombre de clusters K = {}, le coefficient de silhouette est {}, le coefficient d'inertie est {}.".format(K, silhouette_avg, kmeans.inertia_))
#Calcul des largeurs de silhouette :
sample_silhouette_values = silhouette_samples(data, cluster_labels)
```

On peut voir que plus K augmente, plus le coefficient de silhouette diminue et plus le coefficient d'inertie diminue ce qui montre que c'est pour K=2 que le clustering ce passe pour le mieux.
